 Let’s consider two discrete random variables 𝑋 and 𝑌 with the following (joint) probability mass function.

```python
import pandas as pd
import numpy as np


data = [
    # x, y, p_x,y(x,y) = P(X=x, Y=y)
    (-1, 1, 0.06),
    (-1, 2, 0.10),
    (-1, 4, 0.04),
    ( 0, 1, 0.21),
    ( 0, 2, 0.35),
    ( 0, 4, 0.14),
    ( 2, 1, 0.03),
    ( 2, 2, 0.05),
    ( 2, 4, 0.02) 
]

df = pd.DataFrame(data, columns=['x', 'y', 'p_x_y'])
df
```

#### Question 1
 Compute the (marginal) probability mass function of 𝑋 by “summing over the probabilities of 𝑌”.

```python
p_X = df.groupby("x")["p_x_y"].sum()

print(p_X)
```

#### Question 2
Compute the expectation 𝔼(𝑋) of 𝑋.

```python
# print(p_X.index)
# print(p_X.index * p_X)

E_X = (p_X.index * p_X).sum()

print(E_X)
```

#### Question 3
Compute the Variance Var(𝑋) of 𝑋.

```python

# print((p_X.index**2 * p_X).sum())

E_X2 = (p_X.index**2 * p_X).sum() # E(X^2)

Var_X = E_X2 - E_X**2

print(round(Var_X, 4))
```

#### Question 4
Compute the standard deviation SD(𝑋) = sqrt(Var(𝑋)) of 𝑋

```python
SD_X = Var_X**0.5

print(SD_X)
```

#### Question 5
Similarly, compute the (marginal) probability mass function 𝑝𝑌(𝑦), expectation 𝔼(𝑌), variance Var(𝑌), and standard deviation SD(𝑌) of 𝑌.

```python
p_Y = df.groupby('y')["p_x_y"].sum()

print(p_Y) # marginal probability mass function 𝑝𝑌(𝑦)

E_Y = (p_Y.index * p_Y).sum()

print(round(E_Y, 4)) # expectation 𝔼(𝑌)

E_Y2 = (p_Y.index**2 * p_Y).sum() # E(Y^2)

Var_Y = E_Y2 - E_Y**2

print(round(Var_Y, 4)) # variance Var(𝑌)

SD_Y = Var_Y**0.5 

print(round(SD_Y, 4)) # standard deviation SD(𝑌)
```

#### Question 6
Compute the covariance Cov(𝑋, 𝑌) of 𝑋 and 𝑌. 

```python
E_XY = (df["x"] * df["y"] * df["p_x_y"]).sum()

print(round(E_XY)) # expectation of the product 𝔼(𝑋𝑌)

Cov_XY = E_XY - E_X * E_Y

print(round(Cov_XY)) # covariance Cov(𝑋,𝑌)
```

#### Question 7
Compute the correlation Corr(𝑋, 𝑌) of 𝑋 and 𝑌. 

```python
SD_X = Var_X ** 0.5
SD_Y = Var_Y ** 0.5

Corr_XY = Cov_XY / (SD_X * SD_Y)

print(round(Corr_XY)) # correlation Corr(𝑋,𝑌)
```

#### Question 8
Based on Question 7, are random variables 𝑋 and 𝑌 correlated or uncorrelated?

Corr_XY = 0, Therefore 𝑋 and 𝑌 are uncorrelated.
