 Let’s consider two discrete random variables 𝑋 and 𝑌 with the following (joint) probability mass function.

```python
import pandas as pd
import numpy as np


data = [
    # x, y, p_x,y(x,y) = P(X=x, Y=y)
    (-1, 1, 0.06),
    (-1, 2, 0.10),
    (-1, 4, 0.04),
    ( 0, 1, 0.21),
    ( 0, 2, 0.35),
    ( 0, 4, 0.14),
    ( 2, 1, 0.03),
    ( 2, 2, 0.05),
    ( 2, 4, 0.02) 
]

df = pd.DataFrame(data, columns=['x', 'y', 'p_x_y'])
df
```

#### Question 1
 Compute the (marginal) probability mass function of 𝑋 by “summing over the probabilities of 𝑌”.

```python
p_X = df.groupby("x")["p_x_y"].sum()

print(p_X)
```

#### Question 2
Compute the expectation 𝔼(𝑋) of 𝑋.

```python
# print(p_X.index)
# print(p_X.index * p_X)

E_X = (p_X.index * p_X).sum()

print(E_X)
```

#### Question 3
Compute the Variance Var(𝑋) of 𝑋.

```python

# print((p_X.index**2 * p_X).sum())

E_X2 = (p_X.index**2 * p_X).sum() # E(X^2)

Var_X = E_X2 - E_X**2

print(round(Var_X, 4))
```

#### Question 4
Compute the standard deviation SD(𝑋) = sqrt(Var(𝑋)) of 𝑋

```python
SD_X = Var_X**0.5

print(SD_X)
```

#### Question 5
Similarly, compute the (marginal) probability mass function 𝑝𝑌(𝑦), expectation 𝔼(𝑌), variance Var(𝑌), and standard deviation SD(𝑌) of 𝑌.

```python
p_Y = df.groupby('y')["p_x_y"].sum()

print(p_Y) # marginal probability mass function 𝑝𝑌(𝑦)

E_Y = (p_Y.index * p_Y).sum()

print(round(E_Y, 4)) # expectation 𝔼(𝑌)

E_Y2 = (p_Y.index**2 * p_Y).sum() # E(Y^2)

Var_Y = E_Y2 - E_Y**2

print(round(Var_Y, 4)) # variance Var(𝑌)

SD_Y = Var_Y**0.5 

print(round(SD_Y, 4)) # standard deviation SD(𝑌)
```

#### Question 6
Compute the covariance Cov(𝑋, 𝑌) of 𝑋 and 𝑌. 

```python
E_XY = (df["x"] * df["y"] * df["p_x_y"]).sum()

print(round(E_XY,4)) # expectation of the product 𝔼(𝑋𝑌)

Cov_XY = E_XY - E_X * E_Y

print(round(Cov_XY,4)) # covariance Cov(𝑋,𝑌)
```

#### Question 7
Compute the correlation Corr(𝑋, 𝑌) of 𝑋 and 𝑌. 

```python
SD_X = Var_X ** 0.5
SD_Y = Var_Y ** 0.5

Corr_XY = Cov_XY / (SD_X * SD_Y)

print(round(Corr_XY, 4)) # correlation Corr(𝑋,𝑌)
```

#### Response & Suggestions:
Potential data issues: Chicago Taxi Trips 2022
A. Cross-cutting issues (highest importance)

1. Missingness is structural, not random

Why: The city masks locations for privacy (low-volume tracts, trips outside Chicago). Some companies or payment types may not report fare fields at all.
Detect: df.isna().groupby(df['Company']).mean(), and the same by Payment Type, month and area.
Handle: Treat missingness as information. Report which groups it hits before dropping or imputing.
Risk: Silent bias toward certain companies, payment types or neighborhoods.

2. Bias introduced by the cleaning itself

Why: Dropping rows with missing fares or locations removes disproportionate shares of certain groups (for example, cash trips or low-density areas).
Detect: Compare payment mix, hour-of-day, company share and area distribution before and after cleaning.
Handle: Document what was removed and why. Test whether conclusions change under looser or stricter thresholds.
Risk: Conclusions that apply only to the surviving subset.

3. Arbitrary outlier thresholds

Why: Cutoffs such as "under 2 hours" or "under 100 miles" are judgment calls.
Handle: Run a sensitivity check at several thresholds, and keep a log of rows removed per rule.
B. Timestamps

4. Parsing and rounding

Strings need pd.to_datetime(..., format='%m/%d/%Y %I:%M:%S %p'). Timestamps are very likely rounded to 15-minute intervals for privacy, so derived durations are only approximate.

5. Daylight saving time

Times are local and timezone-naive. The spring-forward date has a missing hour that can make derived durations wrong by an hour. The fall-back date has a repeated hour that creates ambiguity and apparent negative durations.
Detect: Compare Trip Seconds against End − Start and look at where the disagreements cluster by date.

6. Logical and range errors

End before start, years outside 2022 (end dates can legitimately spill into Jan 2023), and zero-length trips.
Detect: df[df.start > df.end], df.end.dt.year.value_counts().
C. Trip Seconds and Trip Miles

7. Zeros and extremes

Expect many zeros (cancelled trips, meter glitches), 24-hour trips, and distances in the thousands of miles. Zero miles with nonzero time, and zero time with nonzero miles, are both common.
Handle: Define a plausible range (for example 1 s to a few hours, 0.01 to ~100 miles) and justify it.

8. Implied speed (often overlooked)

Range filters on each column separately miss combinations such as 50 miles in 3 minutes.
Detect: speed = miles / (seconds / 3600), then inspect the upper tail and drop physically impossible speeds (for example above ~80 mph).

9. Duration consistency

Trip Seconds and timestamp difference should agree within the 15-minute rounding tolerance. Large gaps indicate a bad record.
D. Money columns

10. Zero, tiny and absurd values

Zero or one-cent fares, fares in the thousands, and trips where tolls or extras are in the thousands. Taxi fares have a base amount, so anything below it is suspect.
Detect: df['Fare'].value_counts().sort_index().head(20) and .nlargest(20).

11. Cash tips are not recorded

Tips are typically captured only for card or mobile payments. For cash, Tips = 0 means "unknown", not "no tip".
Handle: Analyze tipping only on non-cash payments.
Risk: Badly understated tip rates, and a spike at zero in any tip histogram.

12. Components may not sum to Trip Total

Taxes, surcharges or other fees may not appear in any column.
Detect: Use np.isclose(...) or round to cents. Comparing floats with == produces false mismatches from rounding error.

13. Spikes at fixed amounts

Base fare, flat airport rates and standard surcharges create peaks at specific values. These are real, not errors. Don't "fix" them.
E. Location columns

14. Centroids, not true positions

Pickup and dropoff coordinates are centroids of tracts or community areas, so many trips share identical points and the number of unique coordinates is small.
Risk: Distance computed from coordinates is crude, and maps look gridded. Don't treat them as GPS data.

15. Redundant and inconsistent columns

Lat, lon and the WKT string carry the same information. Parse the WKT and confirm it matches. Keep one representation.

16. Codes stored as floats

Census tracts are 11-digit identifiers and community areas are integers 1–77. NaNs force them to float. Convert to string or nullable Int64.

17. Tract vs community area availability

Often only one of the two is present. Decide which is your location key.
F. Categorical columns

18. Company names

Spelling and case variants, trailing "Inc." or ", LLC", truncated names, and owner-operator entries labeled with ID numbers.
Detect: df['Company'].value_counts().sort_index() and merge near-duplicates.

19. Payment type

Categories such as Unknown, Dispute, No Charge and Prcard. "Unknown" may reflect one company's reporting practice, not a real payment method. Check the company × payment type table before merging categories.
Dispute and No Charge rows usually have unreliable fares.
G. IDs and duplicates

20. Taxi ID is a vehicle, not a driver

Counts per taxi are heavily skewed, and busy taxis are likely shared across drivers and shifts.
Detect: Sort by Taxi ID and start time, and flag overlapping trips. A single taxi cannot be on two trips at once.

21. Hidden duplicates

Trip ID may be unique while the underlying trip is duplicated. Check duplicates on all other columns.
H. Context and technical

22. 2022 context

It was a post-COVID recovery year. Low January volume, seasonality, weather and holidays are real patterns, not data errors. Weekday and hour effects are strong.

23. Memory and dtypes

At ~6.4M rows × 23 columns, the data is about 3 GB in pandas. Use category for company and payment type, downcast numerics, and drop redundant columns early. Consider parquet over CSV.

Check cross-field consistency (speed, duration, money).
Standardize categories.
Apply filters, logging the rows removed per rule.
