#### Question 1a. Mean Estimator:

```python
set.seed(1)

num.of.estimate <- 1000000
sample.means <- numeric(num.of.estimate)
for (i in seq_len(num.of.estimate)) {
n <- 10
x <- rnorm(n, mean = 7, sd = 2)
sample.means[i] <- mean(x)
}
mean(sample.means)
mean_sample_mean <- mean(sample.means)
```

#### Question 1b. Median Estimator:

```python
sample.medians <- numeric(num.of.estimate)
for (i in seq_len(num.of.estimate)) {
n <- 10
x <- rnorm(n, mean = 7, sd = 2)
sample.medians[i] <- median(x)
}

mean_sample_median <- mean(sample.medians)
mean_sample_median
```

Because the normal distribution is symmetric, its mean equals its median,
and the sample median is also unbiased for mu in this case.

#### Question 1c. Mid-Range Estimator:

```python
sample.midrange <- numeric(num.of.estimate)

for (i in seq_len(num.of.estimate)) {
  x <- rnorm(n, mean = 7, sd = 2)
  sample.midrange[i] <- (min(x) + max(x)) / 2
}

mean_sample_midrange <- mean(sample.midrange)
mean_sample_midrange
```

The mid-range is also unbiased for mu

#### Question 1d. UMVUE:

```python
var_mean     <- var(sample.means)
var_median   <- var(sample.medians)
var_midrange <- var(sample.midrange)

data.frame(
  estimator = c("sample mean", "sample median", "mid-range"),
  mean_MC   = c(mean_sample_mean, mean_sample_median, mean_sample_midrange),
  var_MC    = c(var_mean, var_median, var_midrange)
)
```

The sample mean is the best (minimum-variance unbiased) estimator of mu among the three options.

#### Question 2. Unbiased Estimate of Noise Variance sd^2:

```python
set.seed(1)

num.of.repeat <- 10000
sigma.sq.hat <- numeric(num.of.repeat)
for (i in seq_len(num.of.repeat)) {
# Step 1: Simulate design matrix X
n <- 100
d <- 5
X <- cbind(1, matrix(rnorm(n * d), n, d))
# Step 2: Simulate response y
beta <- 1:(d + 1) # True parameter
sigma.sq <- 0.09 # True parameter
y <- X %*% beta + rnorm(n, sd = sqrt(sigma.sq))
# Step 3: Fit linear regression model
data <- data.frame(X, y)
model <- lm(y ~ ., data = data)
# Step 4: Compute variance of residuals
sigma.sq.hat[i] <- var(residuals(model)) # Based on 1 / (n - 1) * sum(...)
}

mean(sigma.sq.hat)
p <- length(beta)
mean(sigma.sq.hat * (n - 1) / (n - p)) # Unbiased for sigma.sq
```

Step 4 tends to underestimate the true value of sd^2

The adjustments was "(n-1)/(n-p)"
