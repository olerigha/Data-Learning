# Neo4J - Py2Neo

```python
from py2neo import Graph, Node, Relationship
import networkx as nx
import matplotlib.pyplot as plt
```

```python
g = Graph("bolt://localhost:7687", auth=("neo4j", "dataengg123"))
```

```python
# delete all the data
g.delete_all()
```

### Nodes  

Create nodes with the Node class. The first argument is the node's label. The remaining arguments are an arbitrary amount of node properties or key-value pairs

```python
nicole = Node("Person", name="Nicole", age=24)
drew = Node("Person", name="Drew", age=20)

mtdew = Node("Drink", name="Mountain Dew", calories=9000)
cokezero = Node("Drink", name="Coke Zero", calories=0)

coke = Node("Manufacturer", name="Coca Cola")
pepsi = Node("Manufacturer", name="Pepsi")

g.create(nicole | drew | mtdew | cokezero | coke | pepsi)
```

### Relationships 

Create relationships between nodes with the Relationship class.

```python
g.create(Relationship(nicole, "LIKES", cokezero))
g.create(Relationship(nicole, "LIKES", mtdew))
g.create(Relationship(drew, "LIKES", mtdew))
g.create(Relationship(coke, "MAKES", cokezero))
g.create(Relationship(pepsi, "MAKES", mtdew))
```

### Visualize

```python
# Create a NetworkX graph
G = nx.Graph()

# Fetch the graph data (all nodes and relationships)
data = g.run("MATCH (n)-[r]->(m) RETURN n, r, m").data()

# Add nodes to the NetworkX graph
for record in data:
    # Add nodes (n and m) to the graph
    G.add_node(record['n']['name'], label=record['n'].labels)
    G.add_node(record['m']['name'], label=record['m'].labels)
    
    # Add edges (relationships) between nodes
    G.add_edge(record['n']['name'], record['m']['name'], type=record['r'].type)

# Draw the graph
plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G)  # Layout for nodes

# Draw the nodes and edges
nx.draw(G, pos, with_labels=True, node_size=3000, node_color="skyblue", font_size=12, font_weight="bold", edge_color="gray")

# Display the title and show the plot
plt.title("Graph of People, Drinks, and Manufacturers")
plt.show()
```

### Queries

Retrieve Cypher query results with Graph.cypher.execute.

```python
query = """
MATCH (person:Person)-[:LIKES]->(drink:Drink)
RETURN person.name AS name, drink.name AS drink
"""

data = g.run(query)

for d in data:
    print(d)
```

### Parameterized Cypher 

Pass parameters to Cypher queries by passing additional key-value arguments to Graph.cypher.execute. Parameters in Cypher are named and are wrapped in curly braces.

```python
query = """
MATCH (p:Person)-[:LIKES]->(drink:Drink)
WHERE p.name = $name
RETURN p.name AS name, AVG(drink.calories) AS avg_calories
"""
data = g.run(query, name="Nicole")

for d in data:
    print(d)
```
