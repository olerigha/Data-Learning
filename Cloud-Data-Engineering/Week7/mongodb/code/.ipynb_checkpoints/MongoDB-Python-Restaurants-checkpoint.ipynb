### Create a Text Index and Search for text

```python
#create a text index for searching
db.restaurants.create_index([('name', pymongo.TEXT)], name='search_index', default_language='english')
```

```python
#find all documents that have the word "Sport" on indexed fields
query = {"$text": {"$search": 'Sport'}}
      
print(db.restaurants.count_documents(query))

for doc in db.restaurants.find(query).limit(3):
    pprint(doc)
```

### Update documents

update_one(),update_many()

In MongoDB, a write operation is atomic on the level of a single document, even if the operation modifies multiple embedded documents within a single document.

When a single write operation modifies multiple documents, the modification of each document is atomic, but the operation as a whole is not atomic and other operations may interleave.However, you can isolate a single write operation that affects multiple documents using the $isolated operator.

```python
#update document 
result = db.restaurants.update_one(
    {"cuisine": "Indian"},
    {
        "$set": {
            "borough": "Manhattan",
            "name": "Mexican Garden"
        },
        "$currentDate": {"lastModified": True}
    }
)
```

```python
#documents that were updated
print (result.matched_count)
print (result.modified_count)
cursor = db.restaurants.find({"name": "Mexican Garden"}).limit(10)
for document in cursor: 
    pprint(document)
```

### Replace documents

To replace the entire document rather than selected fields

replace_one()

```python
#update documents
#After the update, the document only contains the field or fields in the replacement document.
result = db.restaurants.replace_one(
    {"restaurant_id": "41704620"},
    {
        "name": "Mexican Garden",
        "cuisine": "Mexican",
        "address": {
            "coord": [-73.9557413, 40.7720266],
            "building": "1480",
            "street": "2 Avenue",
            "zipcode": "10075"
        }
    }
)
```

```python
#documents that were updated
print (result.matched_count)
print (result.modified_count)

for doc in db.restaurants.find({"name": "Mexican Garden"}).limit(10):
    pprint(doc)
```

### Data Aggregation, Grouping & Sorting 
Documents enter a multi-stage pipeline that transforms the documents into aggregated results

```python
# Groups documents by city and get counts of each sorted in descending order
for doc in db.restaurants.aggregate(
        [ 
            { '$group': { '_id': "$borough", "count": { '$sum': 1 } } },
            { '$sort' : {'count' : -1} }
        ]
    ):
    pprint(doc)
```

```python
# find a list of restaurants located in the Bronx, grouped by cuisines
for doc in db.restaurants.aggregate( 
      [ 
          { '$match': { "borough": "Bronx" } },
          { '$group': { '_id': "$cuisine",  "count": { '$sum': 1 } } },
      ]  ):
    pprint(doc)
```

```python
# The following pipeline uses $match to query the restaurants collection for documents with borough 
# equal to "Queens" and cuisine equal to Brazilian. The _id field contains the distinct zipcode value.
for doc in db.restaurants.aggregate(
   [
     { '$match': { "borough": "Queens", "cuisine": "Brazilian" } },
     { '$group': { "_id": "$address.zipcode" , "count": { '$sum': 1 } } }
   ] ):
    pprint(doc)
```

### Geospatial Queries

Reference: https://docs.mongodb.com/manual/geospatial-queries/

```python
#find all restaurants with lat or long below value
for doc in db.restaurants.find({"address.coord" : {"$lt" : -95.754168}}).limit(3):
    pprint(doc)
```

```python
#create index for geospatial queries
db.restaurants.create_index([("address.coord", GEO2D)])
```

```python
map_center = [-73.961704, 40.662942]  # The location of the query (Longitude, Latitude)
```

```python
#find all restaurants near a point 
query = {"address.coord": SON([("$near", map_center), ("$maxDistance", 0.001)])}
#query = {"address.coord": SON([("$near", map_center), ("$maxDistance", 0.01)])}
#query = {"address.coord": SON([("$near", map_center), ("$maxDistance", 0.1)])}

for doc in db.restaurants.find(query):
    pprint(doc['name'])
    pprint(doc['address'])
```

```python
# Geospatial query using the aggregation pipeline
pipeline = [
    {
        "$geoNear": {
            "near": map_center,  # The reference point for the search
            "distanceField": "dist.calculated",  # Stores the distance in the resulting documents
            "maxDistance": 0.0001,  # Maximum distance in units
            "spherical": True,  # Ensure the use of spherical geometry for the query
            "key": "address.coord"  # Name the 2d index to use; required when the collection has more than one
        }
    },
    {
        "$count": "total_restaurants"  # Count the documents returned by the geo query
    }
]

# Run the aggregation pipeline
results = list(db.restaurants.aggregate(pipeline))

# Print the count
if results:
    print(results[0]['total_restaurants'])  # Get the count of matching documents
else:
    print(0)  # If no results are returned
```

```python
# Define a geospatial query (example using $geoNear)
pipeline = [
    {
        "$geoNear": {
            "near": map_center,  # Reference point for the query (longitude, latitude)
            "distanceField": "dist.calculated",  # Calculate distance and store it in 'dist.calculated'
            "maxDistance": 0.0001,  # Maximum distance in meters
            "spherical": True,  # Use spherical geometry
            "key": "address.coord"  # Name the 2d index to use when the collection has more than one
        }
    }
]

# Run the aggregation pipeline
results = list(db.restaurants.aggregate(pipeline))
#pprint(results)
```

```python
# Extract the coordinates from the query result
coordinates = []
names = []
for result in results:
    coord = result['address']['coord']
    name = result['name']
    coordinates.append(coord)
    names.append(name)

# For visualization, let's also print the coordinates
#print(coordinates)

# MongoDB stores coordinates as [longitude, latitude]; folium expects [latitude, longitude]
center_latlng = [map_center[1], map_center[0]]

# Create a folium map centered at the query location
m = folium.Map(location=center_latlng, zoom_start=20, tiles="OpenStreetMap")

# Add a marker for the query center
folium.Marker(location=center_latlng, popup='NYC City Center', icon=folium.Icon(color='red')).add_to(m)

# Add the coordinates to the map
for coord,name in zip(coordinates,names):
    folium.Marker(
            location=[coord[1], coord[0]],
            popup=name  # display the business name
        ).add_to(m)

# Display the map
display(m)
```

### Polygon Queries with Neighborhoods

The queries above work with points. GeoJSON polygons let us ask which area contains a point, and which restaurants fall inside an area.

Import the neighborhood boundaries (195 polygons) into the `food` database. The path is relative to this notebook's folder.

GeoJSON coordinates are ordered `[longitude, latitude]`.

```python
!mongoimport --db food --collection neighborhoods --drop --file ../data/neighborhoods.json
```

```python
#create a 2dsphere index for GeoJSON geometries
db.neighborhoods.create_index([("geometry", pymongo.GEOSPHERE)], name="geo_index_neighborhoods")
```

#### Which neighborhood contains a point? ($geoIntersects)

```python
point = {"type": "Point", "coordinates": [-73.93414657, 40.82302903]}

neighborhood = db.neighborhoods.find_one(
    {"geometry": {"$geoIntersects": {"$geometry": point}}}
)
print(neighborhood["name"])
```

#### Restaurants inside a neighborhood ($geoWithin)

Restaurant locations are stored as points in `address.coord`, which `$geoWithin` can match against the neighborhood polygon.

```python
query = {"address.coord": {"$geoWithin": {"$geometry": neighborhood["geometry"]}}}

print(db.restaurants.count_documents(query))

for doc in db.restaurants.find(query).limit(3):
    print(doc["name"], "-", doc["cuisine"])
```

#### Restaurants in a sample of neighborhoods

```python
for hood in db.neighborhoods.find().limit(5):
    count = db.restaurants.count_documents(
        {"address.coord": {"$geoWithin": {"$geometry": hood["geometry"]}}}
    )
    print(f"{hood['name']}: {count}")
```

#### Map the neighborhood

```python
# folium.GeoJson reads GeoJSON as [longitude, latitude]; markers need [latitude, longitude]
m = folium.Map(tiles="OpenStreetMap")

boundary = folium.GeoJson(neighborhood["geometry"], name=neighborhood["name"]).add_to(m)
m.fit_bounds(boundary.get_bounds())

for doc in db.restaurants.find(query):
    lng, lat = doc["address"]["coord"]
    folium.Marker(location=[lat, lng], popup=doc["name"]).add_to(m)

display(m)
```
