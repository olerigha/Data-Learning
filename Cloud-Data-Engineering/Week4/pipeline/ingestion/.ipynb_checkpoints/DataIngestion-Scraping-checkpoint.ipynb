# Data Ingestion - Scraping a website

We will scrape the IMDB website using Python and the BeautifulSoup package.

Example URL: 

https://www.imdb.com/search/title/?release_date=2015-01-01,2015-12-31

```python
#Import packages
from time import time, sleep 
from random import randint
import pandas as pd
import re
import requests
from bs4 import BeautifulSoup
import urllib
from urllib.request import urlopen, urlretrieve
from warnings import warn
import matplotlib.pyplot as plt
```

```python
# Lists to store scraped data
names = []
years = []
imdb_ratings = []
votes = []

# Year range to scrape
start_year = 2020
end_year = 2023
```

## Scrape the Website

```python
def fetch_url(url, headers):
    """Fetch the URL and return the response."""
    response = requests.get(url, headers=headers)
    print(f"Fetching: {url}")
    sleep(randint(4, 10))  # Avoid overwhelming the server
    print(response.status_code)
    return response

def parse_vote_count(vote_text):
    """Convert vote count text to an integer."""
    vote_text = vote_text.replace('(', '').replace(')', '').replace('K', '000')
    
    # Remove non-numeric characters except for the decimal point and 'M'
    cleaned_value = ''.join(filter(lambda x: x.isdigit() or x in ['.', 'M'], vote_text))
    
    try:
        if 'M' in vote_text:
            return int(float(cleaned_value.replace('M', '')) * 1_000_000)
        else:
            return int(float(cleaned_value))
    except ValueError as e:
        print(f"Error converting votes to int: {e}")
        return None  # Return None in case of an error
```

```python
# Monitor the number of requests
requests_made = 0

def scrape_movies(year):
    """Scrape movies from a specific year."""
    global requests_made
    
    url = f'https://www.imdb.com/search/title/?release_date={year}-01-01,{year}-12-31'
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/85.0.4183.121 Safari/537.36'
    }

    try:
        response = fetch_url(url, headers)
        requests_made += 1
        
        if requests_made > 72:
            warn('Number of requests was greater than expected.')

        page_html = BeautifulSoup(response.content, 'html.parser')
        mv_containers = page_html.find_all('li', class_='ipc-metadata-list-summary-item')

        for container in mv_containers:
            # Scrape movie details
            name_tag = container.find('h3', class_='ipc-title__text')
            if name_tag:
                names.append(name_tag.get_text(strip=True))

            year_tag = container.find('span', class_=lambda x: x and 'metadata-item' in x)
            if year_tag:
                years.append(year_tag.get_text(strip=True).split('–')[0])

            imdb_tag = container.find('span', class_='ipc-rating-star--rating')
            if imdb_tag:
                imdb_ratings.append(float(imdb_tag.get_text(strip=True)))

            vote_tag = container.find('span', class_=lambda x: x and 'voteCount' in x)
            if vote_tag:
                int_value = parse_vote_count(vote_tag.get_text(strip=True))
                votes.append(int_value)
            else:
                votes.append(None)  # Append None if no votes are present
    except Exception as e:
        print(f'Error occurred: {e}')

# Loop through the range of years and scrape data
for year in range(start_year, end_year + 1):
    scrape_movies(year)

# Print collected data
print(f'Names: {names}')
print(f'Years: {years}')
print(f'IMDB Ratings: {imdb_ratings}')
print(f'Votes: {votes}')
```

## Data Exploration

```python
movie_ratings = pd.DataFrame({'movie': names,
                              'year': years,
                              'imdb': imdb_ratings,
                              'votes': votes
})

movie_ratings.info()
```

```python
movie_ratings = movie_ratings[['movie', 'year', 'imdb', 'votes']]
movie_ratings.head(15)
```

```python
movie_ratings.to_csv("movie_ratings.csv")
```

```python
#adjust the scale to before plotting together
movie_ratings['n_imdb'] = movie_ratings['imdb'] * 10
movie_ratings.head(3)
```

## Visualize

```python
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(18, 5))
ax1, ax2, ax3 = axes

# Plot the IMDB rating histogram
ax1.hist(movie_ratings['imdb'], bins=20, range=(0, 10), color='skyblue', edgecolor='black')
ax1.set_title('IMDB Rating Distribution', fontsize=14)
ax1.set_xlabel('Rating', fontsize=12)
ax1.set_ylabel('Frequency', fontsize=12)
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# Plot the votes histogram
ax2.hist(movie_ratings['votes'], bins=20, range=(0, 100), color='lightgreen', edgecolor='black')
ax2.set_title('Votes Distribution', fontsize=14)
ax2.set_xlabel('Number of Votes', fontsize=12)
ax2.set_ylabel('Frequency', fontsize=12)
ax2.grid(axis='y', linestyle='--', alpha=0.7)

# Plot the normalized IMDB distribution
ax3.hist(movie_ratings['n_imdb'], bins=20, range=(0, 100), histtype='step', color='purple', label='Normalized IMDB')
ax3.set_title('Normalized IMDB Distribution', fontsize=14)
ax3.set_xlabel('Score', fontsize=12)
ax3.set_ylabel('Frequency', fontsize=12)
ax3.legend(loc='upper right')
ax3.grid(axis='y', linestyle='--', alpha=0.7)

# Hide top and right spines for all axes
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()  # Adjust subplot params to give specified padding
plt.show()
```
